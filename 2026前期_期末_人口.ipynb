R8_東松山市の人口（地域別）

```python
# R8_東松山市人口.csvフアイル

csv_files = "/content/mydrive/MyDrive/38_自然言語処理/市民アンケート(2026)/2026/R8_東松山市人口.csv"

# 2. csvフアイルを読み込む
df_population = pd.read_csv(csv_files, encoding="utf-8-sig")
display(df_population)
```

```python
# R8_東松山市人口 地区別人口の棒グラフ

plt.figure(figsize=(10, 6))

sns.barplot(x="地区", y="合計", data=df_population, color="blue")


# ラベル・目盛りの設定
plt.title("R8_東松山市人口 地区別人口",  fontsize=16)
plt.xlabel("地区名", fontsize=16)
plt.ylabel("人口", fontsize=16)
plt.xticks(rotation=60, ha="right", fontsize=16)
plt.yticks(fontsize=16)

plt.grid(axis="y", linestyle="--", alpha=0.7)

# ラベルが見切れないようにレイアウトを自動調整
plt.tight_layout()

plt.show()
```

令和8年の東松山市の人口調査 0歳から105歳

```python
# R8_東松山市年齢別人口.csvフアイル

csv_files = "/content/mydrive/MyDrive/38_自然言語処理/市民アンケート(2026)/2026/R8_東松山市年齢別人口.csv"

# 2. csvフアイルを読み込む
df_age = pd.read_csv(csv_files, encoding="utf-8-sig")
display(df_age)

all_avg_age = (df_age["年齢"] * df_age["合計"]).sum() / df_age["合計"].sum()
print(f"R8_東松山市の平均年齢: {all_avg_age:.1f}歳")

m_avg_age = (df_age["年齢"] * df_age["男性"]).sum() / df_age["男性"].sum()
print(f"R8_東松山市の男性の平均年齢: {m_avg_age:.1f}歳")

f_avg_age = (df_age["年齢"] * df_age["女性"]).sum() / df_age["女性"].sum()
print(f"R8_東松山市の女性の平均年齢: {f_avg_age:.1f}歳")
```

ヒストグラム

```python
# 年齢別の人口のヒストグラム

x_column = "年齢"
y_column = "合計"

# グラフの作成
plt.figure(figsize=(12, 6))

# DataFrameの機能を使って棒グラフを描画
# width=1.0 にすることで棒と棒の隙間をなくし、ヒストグラムのように見せます
df_age.plot.bar(
    x=x_column,
    y=y_column,
    width=1.0,
    ax=plt.gca(), # 現在の図(figure)に描画
    color='skyblue',
    edgecolor='black' # 境界線を黒にして見やすくする
)

# 軸ラベルとタイトルの設定
plt.title('R8 年齢別人口分布 (0歳〜105歳)', fontsize=16)
plt.xlabel('年齢 (歳)', fontsize=16)
plt.ylabel('人口 (人)', fontsize=16)

# X軸の目盛り調整（105歳まで1歳ずつだとラベルが潰れるため、10歳刻みなどで表示）
plt.xticks(ticks=range(0, 106, 10), labels=range(0, 106, 10), rotation=0)

# グリッド線の追加（Y軸のみ）
plt.grid(axis='y', linestyle='--', alpha=0.7)

# グラフの表示
plt.tight_layout()
plt.show()
```
